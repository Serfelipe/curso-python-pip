```python
def calcular_precio(precio,iva=0.16):
    total= precio + (precio * iva)
    return total
calcular_precio(100) #En este caso tome el 0.16 por defecto
calcular_precio(100, 0.19) #En este caso toma el 0.19 y olvide el 0.16 que estaba prestablecido
```

```python
def calcular_promedio(*numeros): #esto quiere decir que pueden haber 10, 20 1000 números por eso el *
    if len(numeros)==0:
        return 0
    total= sum(numeros)
    promedio= total / len(numeros)
    return promedio
calcular_promedio (100,520,85,78,96)
```
